```python
from io import StringIO
import pandas as pd
```

```python
html = "<!doctype html>\n<html lang=\"lv\"><head><meta charset=\"utf-8\"><title>Rīgas dzīvokļi</title></head>\n<body><h1>Rīgas dzīvokļi</h1><p>Fiktīvi sludinājumi.</p>\n<table id=\"apartments\"><caption>Īres piedāvājumi</caption><thead><tr><th scope=\"col\">Rajons</th><th scope=\"col\">Platība, m²</th><th scope=\"col\">Īre, EUR</th></tr></thead>\n<tbody><tr><td>Centrs</td><td>45</td><td>550</td></tr>\n<tr><td>Āgenskalns</td><td>60</td><td>650</td></tr>\n<tr><td>Teika</td><td>38</td><td>420</td></tr>\n<tr><td>Purvciems</td><td>52</td><td>480</td></tr>\n<tr><td>Mežaparks</td><td>70</td><td>750</td></tr></tbody></table></body></html>"
```

## read_html atgriež tabulu sarakstu

```python
tables = pd.read_html(StringIO(html), attrs={"id": "apartments"})
print(len(tables))
```

```python
df = tables[0]
df
```

```python
df.shape
```

```python
df["Īre, EUR"].mean()
```

```python
df["EUR/m²"] = df["Īre, EUR"] / df["Platība, m²"]
df
```

```python
df.sort_values("EUR/m²")
```
