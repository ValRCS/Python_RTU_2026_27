```python
import csv
from io import StringIO
from pathlib import Path

csv_text = "date,category,description,amount_eur\n2026-11-01,food,maize,2.15\n2026-11-01,transport,bilete,1.50\n2026-11-02,food,piens,1.65\n2026-11-03,books,gramata,12.00\n2026-11-04,food,auglis,3.20\n2026-11-05,transport,bilete,1.50\n2026-11-06,household,ziepes,2.90\n2026-11-07,food,kafija,4.10\n2026-11-08,books,bloknots,3.60\n2026-11-09,food,risi,2.40\n"
path = Path('datasets/expenses.csv')
if not path.is_file():
    path = Path('expenses.csv')
if path.is_file():
    csv_text = path.read_text(encoding='utf-8')
rows = list(csv.DictReader(StringIO(csv_text)))
print(len(rows))
```

```python
totals = {}
for row in rows:
    category = row["category"]
    totals[category] = totals.get(category, 0.0) + float(row["amount_eur"])
for category, amount in totals.items():
    print(f"{category}: {amount:.2f} EUR")
print(f"Kopā: {sum(totals.values()):.2f} EUR")
```

```python
import json
json_text = json.dumps(totals, ensure_ascii=False, indent=2)
print(json_text)
assert abs(sum(totals.values()) - 35.0) < 1e-9  # kursa datu kopijai
```

```python
import matplotlib.pyplot as plt
plt.bar(list(totals), list(totals.values()))
plt.title("Izdevumi pa kategorijām")
plt.ylabel("EUR")
plt.tight_layout()
plt.show()
```
