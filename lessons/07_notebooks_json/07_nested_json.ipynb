# Ligzdots JSON — receptes (papildu uzdevums)
Fiktīvs piemērs; interneta pieprasījumi nav vajadzīgi. Datu kopija ir iekļauta šūnā.

```python
import json

text = "{\n  \"recipes\": [\n    {\n      \"name\": \"Kartupeļu zupa\",\n      \"ingredients\": [\n        \"kartupelis\",\n        \"burkāns\",\n        \"sīpols\"\n      ],\n      \"minutes\": 35\n    },\n    {\n      \"name\": \"Cepti kartupeļi\",\n      \"ingredients\": [\n        \"kartupelis\",\n        \"eļļa\",\n        \"sāls\"\n      ],\n      \"minutes\": 25\n    },\n    {\n      \"name\": \"Burkānu salāti\",\n      \"ingredients\": [\n        \"burkāns\",\n        \"ābols\"\n      ],\n      \"minutes\": 15\n    }\n  ]\n}\n"
data = json.loads(text)
print(data.keys())
```

```python
for recipe in data["recipes"]:
    if "kartupelis" in recipe["ingredients"]:
        print(recipe["name"])
```

## Uzdevums
Atlasi receptes, kas pagatavojamas 30 minūtēs. Saskaiti sastāvdaļu biežumu ar vārdnīcu.
